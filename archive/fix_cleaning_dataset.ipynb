```python
import pandas as pd
```

```python
# Read the file
df = pd.read_csv("Procurement KPI Analysis Dataset.csv")
df
```

```python
print(df.isnull().sum())
```

```python
print((df == 0).sum())
print(df.describe())
```

```python
# ดูค่าความเบ้ (skewness) ของทุกคอลัมน์ตัวเลข
df[['Quantity', 'Unit_Price', 'Negotiated_Price', 'Defective_Units']].skew()
```

```python
# Defective_Units skew ก่อนหน้า (1.367734) ข้อมูลนี้ เบ้ขวา (outlier เยอะ) ใช้ median ทนต่อ outlier ดีกว่า mean 
#เช่น มีบาง order defect เยอะมาก (เช่น 1000+) จะทำให้ mean สูงผิดจริง
df['Defective_Units'] = df['Defective_Units'].fillna(df['Defective_Units'].median())
df
```

```python
df['Delivery_Date'].isna().sum()
print(df['Delivery_Date'].isna().sum())
print(df[['Order_Date','Delivery_Date']].dtypes)
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Delivery_Date'] = pd.to_datetime(df['Delivery_Date'])
print(df[['Order_Date','Delivery_Date']].dtypes)
```

```python
df['Lead_Time'] = (
    df['Delivery_Date'] - df['Order_Date']
).dt.days

(df['Lead_Time'] < 0).sum()
```

```python
df[df['Lead_Time'] < 0]
```

```python
df = df[df['Lead_Time'] >= 0] #ลบทิ้ง
```

```python
# Check for missing values
print("Missing values:\n", df.isnull().sum())

# Check the value to zero
print("\nZero values:\n", (df == 0).sum())

# Duplicate data check
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
print(df.describe())
#Show dataset size after cleaning
print("\nDataset shape after cleaning:", df.shape)
```
